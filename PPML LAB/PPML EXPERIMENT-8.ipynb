```python
#Q2.WAP TO FIND THE NUMBER OF LINES ,WORDS AND CHARACTER IN FILE
# Open the file
f = open("sample.txt", "r")

# Read the contents of the file
data = f.read()

# Count lines, words and characters
lines = data.splitlines()
words = data.split()
characters = len(data)

# Display the result
print("Number of lines:", len(lines))
print("Number of words:", len(words))
print("Number of characters:", characters)

# Close the file
f.close()
```

```python
#Q3.WAP TO READ  A WORD FROM KEYBOARD AND SEARCH WHETHER THAT WORD IS AVAILABLE IN THE FILE OR NOT
# Open the file
f = open("sample.txt", "r")

# Read file contents
data = f.read()

# Enter a word to search
word = input("Enter a word to search: ")

# Search the word in the file
if word in data.split():
    print("Word is available in the file")
else:
    print("Word is not available in the file")

# Close the file
f.close()
```

```python
#Q4.WAP TO READ A WORD FROM THE KEYBOARD AND COUNT THE WORD IN THE FILE
# Open the file
f = open("sample.txt", "r")

# Read file contents
data = f.read()

# Enter a word from keyboard
word = input("Enter a word to count: ")

# Count the word in the file
count = data.split().count(word)

# Display the result
print("The word", word, "appears", count, "times in the file")

# Close the file
f.close()
```

```python
#6.WAP to find the longest line in file and display that line.
# First, let's create a sample.txt file so the program can read from it
with open("sample.txt", "w") as f_init:
    f_init.write("Short line\nThis is a much longer line of text in the file\nAnother line")

# Open the file
f = open("sample.txt", "r")

# Read all lines from the file
lines = f.readlines()

# Find the longest line
longest_line = max(lines, key=len)

# Display the longest line
print("Longest line in the file is:")
print(longest_line)

# Close the file
f.close()
```
