```python
#1.WAP to print twin prime numbers between 1 to N.
n = int(input("Enter N: "))

for i in range(2, n + 1):
    # Check if i is prime
    c1 = 0
    for j in range(1, i + 1):
        if i % j == 0:
            c1 += 1

    # Check if i+2 is prime
    c2 = 0
    for j in range(1, i + 3):
        if (i + 2) % j == 0:
            c2 += 1

    if c1 == 2 and c2 == 2 and i + 2 <= n:
        print(i, i + 2)
#2.WAP to find the factorial of a number.
n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial =", fact)
#3.WAP to check the leap year.
year = int(input("Enter a year: "))

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
#4.WAP to check whether the string is symmetrical or palindrome.
s = input("Enter a string: ")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
#5.WAP to print even length words in a string.
s = input("Enter a string: ")

words = s.split()

for word in words:
    if len(word) % 2 == 0:
        print(word)
#6.WAP to remove all duplicates from a given string.
s = input("Enter a string: ")

result = ""

for ch in s:
    if ch not in result:
        result = result + ch

print("String after removing duplicates:", result)
```
