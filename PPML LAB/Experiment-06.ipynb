```python
#Q1.	WAP to demonstrate the Class, Object, Instance, Methods, Constructor and Destructor
class student:
  def m1(self):
    print("m1 in student")

  @classmethod
  def m2(self):
    print("m2 in student")

  @staticmethod
  def m3(): # Removed 'self' parameter
    print("m3 in student")
o1=student()
o1.m1()
student.m2()
o1.m3()
student.m3()
```

```python

#Q3.wap to  demonsteate  the concept of method overloarding and methodoverriding in object oriented programming
#method overloading
class A:
  def add(self,a,b):
    print("a+b=",a+b)
  def add(self,a,b,c):
    print("a+b+c=",a+b+c)
  def add(self,a,b,c,d):
    print("a+b+c+d=",a+b+c+d)
o1=A()
o1.add(10,20,30,40)
#method overriding
class A:
  def sub(self,a,b):
    print("substraction=",a-b)
  def mul(self,a,b):
    print("multiplecation",a*b)
class c(A):
  def mul(self,a,b):
    print("multiplication in class c",a*b)
o1=c();
o1.sub(5,3)
o1.mul(5,3)
o2=A()
o2.mul(2,2)





```
