```python
#Q2.	WAP to triple all numbers in a given list of integers using map()
# Given list
numbers = [1, 2, 3, 4, 5]

# Triple each number using map()
result = list(map(lambda x: x * 3, numbers))

# Display result
print(result)

```

```python


#Q3.	WAP to add three given list using python map and lambda.
L1 = [1, 2, 3, 4, 5]
L2 = [6, 7, 8, 9, 10]
L3 = [5, 4, 3, 2, 1]

# Using map with three lists to add elements element-wise
result = list(map(lambda x,y,z:x+y+z, L1,L2,L3))
print(result)
```

```python
#Q4.WAP TO FIND THE RATIO OF POSITIVE NUMBER ,NEGATIVE NUMBER ,ZERO NUMBER
L1=[1,2,3,-1,-2,0,9,40,-55,0,0,8,-99,-100]
positive_number =list(filter(lambda x:x>0,L1))
negative_number =list(filter(lambda x:x<0,L1))
zero_number =list(filter(lambda x:x==0,L1))
print("positive number",positive_number)
print("negative number",negative_number)
print("zero number",zero_number)
count=[len(positive_number),len(negative_number),len(zero_number)]
t_ele=len(L1)
res=list(map(lambda x:x/t_ele,count))
print("res=",res)
```

```python
#Q5.	WAP to convert a given list of integer and a tuple of integer in a list of string using map().
# Given list of integers
L1 = [1, 2, 3, 4, 5]

# Given tuple of integers
T1 = (6, 7, 8, 9, 10)

# Convert list integers to strings
list_result = list(map(str, L1))

# Convert tuple integers to strings
tuple_result = list(map(str, T1))

# Print results
print("List of strings:", list_result)
print("Tuple of strings:", tuple_result)
```
