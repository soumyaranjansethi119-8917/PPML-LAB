```python
#q1.GRADE SYSTEM
import math

# Enter coefficients
a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

# Calculate discriminant
d = b**2 - 4*a*c

# Find roots
if d > 0:
    root1 = (-b + math.sqrt(d)) / (2*a)
    root2 = (-b - math.sqrt(d)) / (2*a)

    print("Roots are real and different")
    print("Root 1 =", root1)
    print("Root 2 =", root2)

elif d == 0:
    root = -b / (2*a)

    print("Roots are real and equal")
    print("Root 1 = Root 2 =", root)

else:
    real = -b / (2*a)
    imaginary = math.sqrt(-d) / (2*a)

    print("Roots are complex")
    print("Root 1 =", real, "+", imaginary, "i")
    print("Root 2 =", real, "-", imaginary, "i")

#Q2.WAP to enter a 5-digit number and print the digits present at odd locations.
    n = int(input("Enter a five digit number: "))

d1 = n // 10000
d3 = (n // 100) % 10
d5 = n % 10

print("Digits at odd locations:", d1, d3, d5)

#Q3.wap to check the string is palindrome or not?
s=input("enter the string:")
rev=s[::-1]
if(rev==s):
    print("palindrome:")
else:
    print("not palindrome:")
#q4.WAP to enter a number and check whether the number is perfect number or not.
    n = int(input("Enter a number: "))

sum = 0

for i in range(1, n):
    if n % i == 0:
        sum = sum + i

if sum == n:
    print(n, "is a Perfect Number")
else:
    print(n, "is Not a Perfect Number")
#Q5.5.	WAP to enter a string and print it in reverse also print the no of vowels and consonant in it.
    s = input("Enter a string: ")

# Reverse the string
print("Reverse:", s[::-1])

vowels = 0
consonants = 0

for ch in s:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Number of vowels:", vowels)
print("Number of consonants:", consonants)

```
