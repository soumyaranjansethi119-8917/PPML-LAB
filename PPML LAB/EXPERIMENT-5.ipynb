```python
#Q1.WAP to input two dictionaries and print the values by merging the two dictionaries.
# Enter two dictionaries
dict1 = eval(input("Enter first dictionary: "))
dict2 = eval(input("Enter second dictionary: "))

# Merge the two dictionaries
merged_dict = {**dict1, **dict2}

# Print the merged dictionary
print("Merged Dictionary:", merged_dict)

# Print the values
print("Values:", merged_dict.values())
#q2.WAP to create a dictionary and print the key which has the maximum unique value.
d = {}

n = int(input("Enter number of elements: "))

for i in range(n):
    key = input("Enter key: ")
    value = int(input("Enter value: "))
    d[key] = value

max_key = max(d, key=d.get)

print("Dictionary:", d)
print("Key with maximum value:", max_key)
print("Maximum value:", d[max_key])
#3.	WAP to enter a dictionary and remove the duplicate value inside the dictionary.
# Enter a dictionary
d = eval(input("Enter a dictionary: "))

# Remove duplicate values
new_dict = {}
for key, value in d.items():
    if value not in new_dict.values():
        new_dict[key] = value

print("Dictionary after removing duplicate values:", new_dict)
#4.	WAP to enter a set and copy the content of the set into a new set one element at a time.
# Enter a set
s = eval(input("Enter a set: "))

# Create a new empty set
new_set = set()

# Copy elements one by one
for i in s:
    new_set.add(i)

print("Original set:", s)
print("New set:", new_set)


#5.	WAP to enter two sets and perform all the set operations on it.
# Enter two sets
set1 = set(map(int, input("Enter elements of first set: ").split()))
set2 = set(map(int, input("Enter elements of second set: ").split()))

print("Set 1:", set1)
print("Set 2:", set2)

# Union
print("Union:", set1 | set2)

# Intersection
print("Intersection:", set1 & set2)

# Difference
print("Set 1 - Set 2:", set1 - set2)
print("Set 2 - Set 1:", set2 - set1)

# Symmetric Difference
print("Symmetric Difference:", set1 ^ set2)

# Subset
print("Set 1 is subset of Set 2:", set1 <= set2)
print("Set 2 is subset of Set 1:", set2 <= set1)

# Superset
print("Set 1 is superset of Set 2:", set1 >= set2)
print("Set 2 is superset of Set 1:", set2 >= set1)

# Disjoint
print("Sets are disjoint:", set1.isdisjoint(set2))
#6.	WAP to enter two different sets with string elements combine both the sets remove any duplicates are present, print the new set.













```
