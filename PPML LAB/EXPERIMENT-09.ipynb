```python
#1.	WAP for ndarray Object, Indexing, and Slicing.
import numpy as np

# Creating an ndarray
arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)

# Indexing
print("First element:", arr[0])
print("Third element:", arr[2])

# Slicing
print("First three elements:", arr[0:3])
print("Last two elements:", arr[3:5])

```

```python
#2.	WAP for Data Types and Structures in NumPy.
import numpy as np

# Creating arrays with different data types
a = np.array([10, 20, 30], dtype=int)
b = np.array([1.5, 2.5, 3.5], dtype=float)
c = np.array([1, 0, 1], dtype=bool)

# Displaying arrays and data types
print("Integer Array:", a)
print("Data Type:", a.dtype)

print("Float Array:", b)
print("Data Type:", b.dtype)

print("Boolean Array:", c)
print("Data Type:", c.dtype)

# Creating a 2D array
d = np.array([[1, 2], [3, 4]])

print("Two Dimensional Array:")
print(d)
```

```python
#3.	WAP for NumPy Array Properties and Functions.
import numpy as np

# Creating an array
arr = np.array([[10, 20, 30], [40, 50, 60]])

# Array properties
print("Array:", arr)
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data Type:", arr.dtype)

# NumPy functions
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Sorted Array:")
print(np.sort(arr, axis=None))
```

```python
#4.	WAP for Statistical Operations and Broadcasting on Arrays.
import numpy as np

# Creating an array
arr = np.array([10, 20, 30, 40, 50])

# Statistical operations
print("Array:", arr)
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Variance:", np.var(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))

# Broadcasting
a = np.array([10, 20, 30])
b = 5

print("Original Array:", a)
print("After Broadcasting:", a + b)
```

```python
#5.	WAP for Saving and Loading Arrays.
import numpy as np

# Creating an array
arr = np.array([10, 20, 30, 40, 50])

# Saving the array into a file
np.save("myarray.npy", arr)

print("Array saved successfully")

# Loading the array from the file
new_arr = np.load("myarray.npy")

print("Loaded Array:", new_arr)
```
