```python
# Q1. Arithmetic Operations

a = int(input("Enter the 1st value: "))
b = int(input("Enter the 2nd value: "))

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)

if b != 0:
    print("Division =", a / b)
    print("Floor Division =", a // b)
else:
    print("Division is not possible")

print("Exponentiation =", a ** b)

# Q2. Area of Circle

r = float(input("Enter the radius of the circle: "))
area = 3.14 * r ** 2
print("Area of the circle =", area)

# Q3. Area of Triangle

s1 = float(input("Enter the 1st side: "))
s2 = float(input("Enter the 2nd side: "))
s3 = float(input("Enter the 3rd side: "))

if s1 + s2 > s3 and s1 + s3 > s2 and s2 + s3 > s1:

    s = (s1 + s2 + s3) / 2

    area = (s * (s - s1) * (s - s2) * (s - s3)) ** 0.5

    print("Area of the triangle =", area)

else:
    print("Invalid triangle")
    
# Q3. Number Conversion

num = int(input("Enter an integer: "))

print("Binary =", bin(num)[2:])
print("Octal =", oct(num)[2:])
print("Hexadecimal =", hex(num)[2:])

# Q4. Roots of Quadratic Equation

import math

a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

if a == 0:
    print("It is not a quadratic equation")

else:
    d = b ** 2 - 4 * a * c

    if d > 0:
        root1 = (-b + math.sqrt(d)) / (2 * a)
        root2 = (-b - math.sqrt(d)) / (2 * a)

        print("Roots are real and different")
        print("Root 1 =", root1)
        print("Root 2 =", root2)

    elif d == 0:
        root = -b / (2 * a)

        print("Roots are real and equal")
        print("Root 1 = Root 2 =", root)

    else:
        real = -b / (2 * a)
        imaginary = math.sqrt(-d) / abs(2 * a)

        print("Roots are complex")
        print("Root 1 =", complex(real, imaginary))
        print("Root 2 =", complex(real, -imaginary))
        
# Q6. Meters to Kilometers

a = float(input("Enter the distance in meters: "))

b = a / 1000

print("Distance in kilometers =", b)
```
